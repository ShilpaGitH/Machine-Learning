```python
df = pd.DataFrame(iris.data,columns=iris.feature_names)
df.head(5)
```

```python
iris.target_names
```

```python
df[df.target == 2].head()
```

```python
df['Flower_name'] =  df.target.apply(lambda x:iris.target_names[x])
```

Code	Meaning

df['Flower_name'] -	Create a new column named Flower_name

df.target -	Access the target column (flower class numbers)

.apply() -	Apply a function to every value in the target column

lambda x: -	Take each target value and temporarily store it in x

iris.target_names[x] -	Get the flower name corresponding to the target number

=	Assign the resulting flower names to the new column

#### Train Using Support Vector Machine (SVM)

```python
from sklearn.model_selection import train_test_split
```

```python
df.head()
```

```python
X = df.drop(['target','Flower_name'], axis='columns')
X.head()
```

```python
y = df.target
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

Code	Meaning

train_test_split()	Split the dataset into training data and testing data

X	Input features (independent variables)

y	Target values (dependent variable)

test_size=0.2	Allocate 20% of the data for testing and 80% for training

X_train	Store training input features

X_test	Store testing input features

y_train	Store training target values

y_test	Store testing target values

=	Assign the split data to the respective variables

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.svm import SVC
```

```python
model = SVC()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict([[5.1,3.2,6.5,2.1]])
```

### Tune parameters


#### 1. Regularization (C)

```python
model_C = SVC(C=1)
model_C.fit(X_train, y_train)
model_C.score(X_test, y_test)
```

```python
model_C = SVC(C=10)
model_C.fit(X_train, y_train)
model_C.score(X_test, y_test)
```

#### 2. Gamma

```python
model_g = SVC(gamma=10)
model_g.fit(X_train, y_train)
model_g.score(X_test, y_test)
```

#### 3. Kernel

#### kernel='linear' → Find a straight-line decision boundary.

Kernel	Meaning

linear -	Separates classes using a straight line (or hyperplane).

poly -	Separates classes using a polynomial (curved) boundary.

rbf -	Separates classes using a flexible, non-linear boundary.

sigmoid -	Uses a sigmoid-based decision function to separate classes.

```python
model_linear_kernal = SVC(kernel='linear') ## kernel='linear'	Use a linear kernel to separate different classes using a straight line (or hyperplane)
model_linear_kernal.fit(X_train, y_train)
```

```python
model_linear_kernal.score(X_test, y_test)
```
