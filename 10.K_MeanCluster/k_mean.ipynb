```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = { 'Marks' : [10, 12, 15, 40, 42, 45, 80, 82, 85]}

df = pd.DataFrame(data)
df
```

```python
X = df[['Marks']]
X
```

### Apply Elbow Method

```python
wcss = []

for k in range(1, 6):

    model = KMeans(n_clusters=k, random_state=42, n_init=10)

    model.fit(X)

    wcss.append(model.inertia_)
```

```python
wcss = []

for k in range(1,6):
    model = KMeans(n_clusters=k)
    model.fit(X)
    wcss.append(model.inertia_)
```

Code	Meaning

wcss = [] -	Create an empty list to store SSE values

range(1, 6) -	Try K values from 1 to 5

KMeans(n_clusters=k) -	Create K clusters

model.fit(X) -	Train the clustering model

model.inertia_ -	Calculate total WCSS (SSE)

wcss.append() -	Store the WCSS value in the list

Important: model.inertia_ automatically calculates the total SSE across all clusters.

For example, when K=2:
\[
WCSS = SSE(C1)+SSE(C2)
\]

```python
print(wcss)
```

```python
plt.plot(range(1, 6), wcss, marker='o')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (SSE)')

plt.title('Elbow Method')

plt.show()
```

```python
model = KMeans(n_clusters=3)

model.fit(X)

print(model.cluster_centers_)
print(model.labels_)
```

```python
model = KMeans(n_clusters=3, random_state=42, n_init=10)
model.fit(X)

print(model.labels_)
```

n_init=10 → Try 10 initializations.



random_state=42 → Make the random initialization reproducible.
