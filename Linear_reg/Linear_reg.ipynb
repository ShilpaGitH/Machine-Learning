```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("homeprices.csv")
df
```

```python
df.info
```

```python
%matplotlib inline
plt.xlabel('Area')
plt.ylabel('Price')
plt.scatter(df.area,df.price, color='red', marker="*")
plt.show()
```

```python
%matplotlib inline
plt.title("Home Price", color='red', size=20, fontstyle= 'italic', fontweight='bold')
plt.xlabel('Area',size=20)
plt.ylabel('Price', size=20)
plt.scatter(df.area,df.price, color='purple', marker='*', s=200)
plt.show()
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[3300]])
```

```python
new_area = pd.DataFrame({'area': [3300]})

prediction = reg.predict(new_area)

print(prediction)
```
