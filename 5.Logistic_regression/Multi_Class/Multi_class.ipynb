```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_digits
```

```python
digit = load_digits()
```

```python
dir(digit)
```

Attribute	Meaning

DESCR -	Description of the dataset

data -	Actual input data (features)

feature_names - Names of the features

frame -	Data in DataFrame format, if available

images -	Images of handwritten digits

target -	Actual labels (0–9)

target_names -	Available digit classes (0–9)

```python
plt.gray
for i in range(5):
    plt.matshow(digit.images[0])
```

```python
digit.target[0:5]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(digit.data,digit.target, test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
print(digit.data.shape)
print(digit.target.shape)
print(digit.images.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

3. What does max_iter mean?

- 100 iterations = 100 learning attempts.
- 1000 iterations = 1000 learning attempts.

```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
plt.matshow(digit.images[6])
```

```python
model.predict([digit.data[6]])
```

```python
model.predict(digit.data[0:5])
```

```python
y_predicted = model.predict(X_test)
```

```python
y_predicted
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test,y_predicted)
cm
```

```python
import seaborn as sn
plt.figure(figsize = (10,7))
sn.heatmap(cm, annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```
